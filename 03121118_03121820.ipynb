#### **Υλοποίηση**
Παρακάτω παρατείθετε η υλοποίηση του απαραίτητου κώδικα για το ερώτημα. Στο τέλος της ενότητας υπάρχουν τα συμπεράσματα τα οποία εξάγουμε απο την εκτέλεση του.

```pyspark
spark_query_1_1 = SparkSession \
    .builder \
    .appName("Query 1: Execution with DataFrames") \
    .config("spark.executor.instances", 4) \
    .config("spark.executor.cores", 1) \
    .config("spark.executor.memory", "2g") \
    .getOrCreate()

crime_df.show(1)  # force loading of df to stabilize results

start_time = time.time()
agg_assault = crime_df.filter(crime_df.crime_code_description.contains("AGGRAVATED ASSAULT"))
grouped = agg_assault.withColumn(
    "age_group",
    when(col("victim_age") < 18, "Children")
    .when((col("victim_age") >= 18) & (col("victim_age") <= 24), "Young Adults")
    .when((col("victim_age") >= 25) & (col("victim_age") <= 64), "Adults")
    .when(col("victim_age") > 64, "Elderly")
)
result = grouped.groupBy("age_group").count()
result = result.orderBy(col("count").desc())

result.explain(mode="formatted")
result.show()

end_time = time.time()
time_df_without_udf = end_time - start_time 

print("Dataframe without UDF execution time: ", time_df_without_udf)

# clean variables
del agg_assault, grouped, result, start_time, end_time
```

```pyspark
# implementation with DataFrames (with UDF)
spark_query_1_2 = SparkSession \
    .builder \
    .appName("Query 1: Execution with DataFrames (With UDF") \
    .config("spark.executor.instances", 4) \
    .config("spark.executor.cores", 1) \
    .config("spark.executor.memory", "2g") \
    .getOrCreate()


def age_to_group(age):
    if age < 18:
        return "Childs"
    if 18 <= age <= 24:
        return "Young Adults"
    if 25 <= age <= 64:
        return "Adults"
    return "Elderly"


age_to_group_udf = udf(age_to_group, StringType())

start_time = time.time()

agg_assault = crime_df.filter(crime_df.crime_code_description.contains("AGGRAVATED ASSAULT"))

grouped = agg_assault.withColumn(
    "age_group",
    age_to_group_udf(col("victim_age"))
)

result = grouped.groupBy("age_group").count()
result = result.orderBy(col("count").desc())

result.explain(mode="formatted")
result.show()

end_time = time.time()
time_df_with_udf = end_time - start_time 

print("Dataframe with UDF execution time: ", time_df_with_udf)

# clean variables
del agg_assault, grouped, result, start_time, end_time
```

```pyspark
# implementation with RDD
spark_query_1_3 = SparkSession \
    .builder \
    .appName("Query 1: Execution with RDD") \
    .config("spark.executor.instances", 4) \
    .config("spark.executor.cores", 1) \
    .config("spark.executor.memory", "2g") \
    .getOrCreate()


def age_to_group(age):
    if age < 18:
        return "Childs"
    if 18 <= age <= 24:
        return "Young Adults"
    if 25 <= age <= 64:
        return "Adults"
    return "Elderly"


crime_rdd = crime_df.rdd

start_time = time.time()

agg_assault = crime_rdd.filter(lambda row: row.crime_code_description == "AGGRAVATED ASSAULT")
age_group = agg_assault.map(lambda row: (age_to_group(row.victim_age), 1))
counts = age_group.reduceByKey(lambda a, b: a + b)
sorted_counts = counts.sortBy(lambda x: x[1], ascending=False)
sorted_counts.count()

end_time = time.time()
time_rdd = end_time - start_time

for group, count in sorted_counts.collect():
    print(f"{group}: {count}")

print("RDD execution time:", time_rdd)

# clean variables
del agg_assault, age_group, counts, sorted_counts, start_time, end_time
```

```pyspark
from pyspark.sql.functions import radians, sin, cos, acos, lit, min, asc, desc

# Implementation with DataFrame (2 executors, 1 core, 2GB memory)
spark_query_4_1_1 = (
    SparkSession.builder
    .appName("Query 4: Nearest Police Station (DataFrame)")
    .config("spark.executor.instances", 2)
    .config("spark.executor.cores", 1)
    .config("spark.executor.memory", "2g")
    .getOrCreate()
)

start_time = time.time()

# Rename columns to avoid ambiguity during join
police_renamed = police_stations_df.select(
    col("division").alias("station_division"),
    col("latitude").alias("p_lat"),
    col("longitude").alias("p_long")
)

# Remove Null Island
crime_filtered = crime_df.filter(
    (col("latitude") != 0.0) & 
    (col("longitude") != 0.0)
).select(
    "division_record_number", 
    "latitude", 
    "longitude"
)

# Every crime connects with every police station
joined_df = crime_filtered.crossJoin(police_renamed) # we use broadcast because police_renamed is short

# The Haversine formula is used to calculate the shortest distance between two points on a sphere, given their latitudes and longitudes
joined_df = joined_df.withColumn(
    "distance",
    acos(
        sin(radians(col("latitude"))) * sin(radians(col("p_lat"))) + 
        cos(radians(col("latitude"))) * cos(radians(col("p_lat"))) * cos(radians(col("p_long")) - radians(col("longitude")))
    ) * lit(6371.0)
)

# Separate the rows based on the Crime ID
window_spec = Window.partitionBy("division_record_number").orderBy(col("distance").asc())

ranked_df = joined_df.withColumn("rank", row_number().over(window_spec)) # the station with the shortest distance gets Rank 1, the second Rank 2...

# Only keep the station with the shortest distance from each crime
nearest_df = ranked_df.filter(col("rank") == 1)

# Group by Station and calculate average distance and "#"
final_result = (
    nearest_df
    .groupBy("station_division")
    .agg(
        avg("distance").alias("average_distance"),
        count("*").alias("#")
    )
    .orderBy(col("#").desc())
)

# Show results
final_result.show(truncate=False)
final_result.explain(mode="formatted")

execution_time_1 = time.time() - start_time
print("Query 4 (DataFrame) execution time (1 core, 2GB): ", execution_time_1)

del police_renamed, crime_filtered, joined_df, ranked_df, nearest_df, final_result, start_time
```

## **5. Συσχέτιση Εισοδήματος και Εγκληματικότητας**

#### **Υλοποίηση**
Παρακάτω παρατείθετε η υλοποίηση του απαραίτητου κώδικα για το ερώτημα. Στο τέλος της ενότητας υπάρχουν τα συμπεράσματα, τα οποία εξάγουμε απο την εκτέλεση του.

```pyspark
QUERY5_CONFIGS = {
    "2x4x8": {
        "executors": 2,
        "cores": 4,
        "memory": "8g"
    },
    "4x2x4": {
        "executors": 4,
        "cores": 2,
        "memory": "4g"
    },
    "8x1x2": {
        "executors": 8,
        "cores": 1,
        "memory": "2g"
    }
}
```

```pyspark
def run_query_5(config_key):
    config = QUERY5_CONFIGS[config_key]

    spark_query_5 = (
        SparkSession.builder
        .appName(f"Query5_{config_key}")
        .config("spark.executor.instances", config["executors"])
        .config("spark.executor.cores", config["cores"])
        .config("spark.executor.memory", config["memory"])
        .config("spark.serializer", KryoSerializer.getName)
        .config("spark.kryo.registrator", SedonaKryoRegistrator.getName)
        .config("sedona.global.index", "true")
        .getOrCreate()
    )
    
    SedonaRegistrator.registerAll(spark)
    
    # Loading GeoJSON file of LA_Census_Blocks
    blocks_df = (
        SedonaContext.create(spark_query_5).read.format("geojson")
        .option("multiLine", "true")
        .load(CENSUS_BLOCKS_2020_GEOJSON_URL)
        .selectExpr("explode(features) as features")
        .select("features.*")
    )
    
    blocks_df.show(1)  # force execution of loading
    
    start_time = time.time()  # start timing after initial loading
    
    flattened_df = blocks_df.select(
        [
         col(f"properties.{col_name}").alias(col_name)  # e.g., renames properties.POP20 to just POP20
         for col_name in blocks_df.schema["properties"].dataType.fieldNames()
        ]
        + ["geometry"]
    ).drop("properties").drop("type")
    
    census_df = (
        flattened_df
        .withColumn("POP20", col("POP20").cast(IntegerType()))
        .withColumn("ZCTA20", col("ZCTA20").cast(StringType()))
        .filter(col("COMM").isNotNull()) 
        .select("COMM", "ZCTA20", "POP20", "geometry")
    )
    
    # Inner Join: keeps rows only if the Zip Code exists in both tables
    census_income_df = census_df.join(
        income_dy_ZIP_2021, 
        census_df.ZCTA20 == income_dy_ZIP_2021.ZIP.cast("string"), 
        "inner"
    )
    
    crime_filtered = (
        crime_df
        .withColumn("year", year(to_date(col("occurrence_date"), "yyyy MMM dd hh:mm:ss a")))  # extracts only the year
        .filter((col("year") >= 2020) & (col("year") <= 2021))  # 2-year period
        .filter((col("latitude") != 0.0) & (col("longitude") != 0.0))  # remove Null Island 
        # Create Geometry Point from Lat/Lon using Sedona
        .withColumn("crime_point", expr("ST_Point(cast(longitude as Decimal(24,20)), cast(latitude as Decimal(24,20)))"))
        .select("division_record_number", "crime_point")
    )
    
    spatial_join_df = (
        crime_filtered.alias("c")
        .join(census_income_df.alias("b"), expr("ST_Contains(b.geometry, c.crime_point)"))  # does geometry hold crime_point
        .select("b.COMM", "c.division_record_number")
    )
    
    # Counting the crimes
    crime_counts_per_comm = spatial_join_df.groupBy("COMM").count().withColumnRenamed("count", "total_crimes_2yrs")
    
    comm_stats = (
        census_income_df
        .groupBy("COMM")
        .agg(
            _sum("POP20").alias("total_pop"),
            # Weighted Income = Sum(Income * Population) / Sum(Population)
            (_sum(col("median_income") * col("POP20")) / _sum("POP20")).alias("avg_income")
        )
        .filter(col("total_pop") > 0) 
    )
    
    # Money vs. Crime
    final_df = (
        comm_stats
        .join(crime_counts_per_comm, "COMM", "left")  # we keep communities even if they had zero crimes 
        .fillna(0, subset=["total_crimes_2yrs"])  # if total_crimes_2yrs = null, it means 0 crimes
        # Annual Crime Rate = (Total Crimes / 2 years) / Population
        .withColumn("crime_rate", (col("total_crimes_2yrs") / 2.0) / col("total_pop"))
        .select("COMM", "avg_income", "crime_rate")
    )

    print(f"Total Communities Analyzed: {final_df.count()}")

    corr_all = final_df.stat.corr("avg_income", "crime_rate")
    print(f"Correlation (All Areas): {corr_all:.4f}")

    top_10_rich = final_df.orderBy(desc("avg_income")).limit(10)
    corr_rich = top_10_rich.stat.corr("avg_income", "crime_rate")
    print(f"Correlation (Top 10 Richest): {corr_rich:.4f}")

    top_10_poor = final_df.orderBy(asc("avg_income")).limit(10)
    corr_poor = top_10_poor.stat.corr("avg_income", "crime_rate")
    print(f"Correlation (Top 10 Poorest): {corr_poor:.4f}")

    execution_time = time.time() - start_time
    print("Query 5 (DataFrame) execution time", config_key, ": ", execution_time)

    print("\nOptimizer Strategy for Spatial Join:")
    spatial_join_df.explain(mode="simple")

    del blocks_df, flattened_df, census_df, census_income_df
    del crime_filtered, spatial_join_df, crime_counts_per_comm, comm_stats
    del final_df, top_10_rich, top_10_poor

    return execution_time
```

#### **Συμπεράσματα**

* Όσον αφορά το execution time, διαπιστώνουμε ότι **η υλοποίηση με 8 executors × 1 core, 2GB memory είναι η γρηγορότερη, ενώ αυτή με 2 executors × 4 cores, 8GB memory παίρνει τον περισσότερο χρόνο**. Τα αποτελέσματα αυτά συνάδουν απόλυτα με το γεγονός πως **ένα Spatial Join επωφελείται περισσότερο από περισσότερους executors, παρά από περισσότερους cores**. Με αυτόν τον τρόπο πιθανώς επιτυγχάνεται καλύτερη παραλληλοποίηση. 

* Το *correlation number* είναι ένας αριθμός ο οποίος μας δείχνει την σχέση μεταξύ δύο μεγεθών ("avg_income", "crime_rate"). Θετικός αριθμός θα προκύψει άμα και τα δύο μεγέθη αυξάνονται (ή μειώνονται) μαζί, ενώ σε αντίθετη περίπτωση, ένας αρνητικός αριθμός προδίδει την ύπαρξη μιας "αντίστροφης" σχέσης μεταξύ των μεγεθών. Προφανώς, άμα το correlation number είναι ίσο με μηδέν, αυτό σημαίνει ότι ο ένας παράγοντας δεν επηρεάζει με κανέναν τρόπο τον άλλον. Τα αποτελέσματα της υλοποίησης συμφωνούν απόλυτα με αυτή τη λογική:

  • **Correlation (All Areas): -0.3096** : Αυτός ο αριθμός μας δείχνει ότι σε γενικές γραμμές, η οικονομική σταθερότητα συνδέεται με την μείωση του εγκλήματος. Είναι κάτι το οποίο βγάζει απόλυτο νόημα και στον πραγματικό κόσμο, καθώς υψηλότερο εισόδημα συνεπάγεται συνήθως μια πιο ασφαλής περιόχη, αν και όχι άτρωτη από εγκλήματα. Ο αριθμός δε βρίσκεται κοντά στο -1, οπότε αυτό σημαίνει ότι το εισόδημα δεν είναι ο μόνος παράγοντας που επηρεάζει την εγκληματική δραστηριότητα. 

  • **Correlation (Top 10 Richest): 0.0936** : Για τις δέκα πιο πλούσιες περιοχές, ο αριθμός βρίσκεται πολύ κοντά στο μηδέν: αυτό σημαίνει ότι το εισόδημα σταματάει να παίζει μεγάλο ρόλο στο επίπεδο ασφάλειας. Τα περισσότερα χρήματα δεν ισοδυναμούν αναγκαστικά με λιγότερα εγκλήματα, αλλά η σχέση που τα συνδέει γίνεται πιο ασθενής.

  • **Correlation (Top 10 Poorest): -0.3701** : Στις δέκα πιο φτωχές περιοχές, η εγκληματικότητα και το εισόδημα συνδέονται με άμεσο τρόπο. Ακόμη και οι πιο μικρές αλλαγές στην οικονομική κατάσταση μιας περιοχής μπορεί να έχουν σημαντική επίδραση στον βαθμό εγκληματικότητας, με την έλλειψη σταθερότητας να παίζει σημαντικό ρόλο σε αυτό. Το νούμερο αυτό υποδεικνύει πόσο σημαντική είναι η ανάπτυξη τέτοιων περιοχών χαμηλού εισοδήματος, καθώς ακόμη και η πιο μικρή επένδυση θα μπορούσε να βοηθήσει τη μείωση του εγκλήματος.


* Τέλος, όσον αφορά τη στρατηγική join που χρησιμοποιείται, στρατηγικές όπως οι *SortMergeJoin* και *ShuffleHashJoin* δε υποστηρίζουν συνθήκες που δεν είναι ισότητα. Για το Spatial Join, η στρατηγική που επιλέχθηκε είναι η **RangeJoin**, η οποία έχει ειδικά σχεδιαστεί για να χειρίζεται ζητήματα με *ST_Contains*. Η *RangeJoin* ανιχνεύει σημεία μέσα σε πολύγωνα χωρίς να αυξήσει την ακρίβεια του προγράμματος παραπάνω από το φυσιολογικό, καθιστώντας την ιδανική για ένα γεωμετρικό πρόβλημα. Αξίζει να σημειωθεί, επίσης, ότι ο optimizer εντόπισε πως το dataset με τα εισοδήματα είναι μάλλον σχετικά μικρό, επιλέγοντας έτσι τη στρατηγική **BroadcastHashJoin** για να συνδέσει προηγουμένως τα δεδομένα του εισοδήματος με τα Census Βlocks.
